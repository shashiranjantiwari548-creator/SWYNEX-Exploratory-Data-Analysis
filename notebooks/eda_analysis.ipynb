```python
from pathlib import Path

Path("../outputs").mkdir(parents=True, exist_ok=True)

print("Output folder ready.")
```

```python
rating_counts = df["rating"].value_counts()

print(rating_counts)
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().index
)

plt.title("Netflix Content Distribution by Rating")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")
plt.tight_layout()

plt.savefig("../outputs/rating_distribution.png", dpi=300)

plt.show()
```

```python
rating_percentage = (
    df["rating"].value_counts(normalize=True) * 100
).round(2)

print(rating_percentage)
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

df["added_year"] = df["date_added"].dt.year

added_year_counts = df["added_year"].value_counts().sort_index()

print(added_year_counts)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    added_year_counts.index,
    added_year_counts.values,
    marker="o"
)

plt.title("Netflix Titles Added to the Platform by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig("../outputs/content_added_by_year.png", dpi=300)

plt.show()
```

```python
country_counts = (
    df["country"]
    .dropna()
    .loc[lambda x: x != "Not Specified"]
    .str.split(", ")
    .explode()
    .value_counts()
)

print(country_counts.head(15))
```

```python
plt.figure(figsize=(10, 7))

country_counts.head(15).sort_values().plot(kind="barh")

plt.title("Top 15 Countries by Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.tight_layout()

plt.savefig("../outputs/top_15_countries.png", dpi=300)
plt.show()
```

```python
genre_counts = (
    df["listed_in"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
)

print(genre_counts.head(15))
```

```python
plt.figure(figsize=(10, 7))

genre_counts.head(15).sort_values().plot(kind="barh")

plt.title("Top 15 Netflix Content Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.tight_layout()

plt.savefig("../outputs/top_15_genres.png", dpi=300)
plt.show()
```

```python
movies = df[df["type"] == "Movie"].copy()

movies["duration_minutes"] = (
    movies["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

print(movies["duration_minutes"].describe())
```

```python
plt.figure(figsize=(10, 6))

plt.hist(
    movies["duration_minutes"].dropna(),
    bins=30
)

plt.title("Distribution of Netflix Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.tight_layout()

plt.savefig("../outputs/movie_duration_distribution.png", dpi=300)
plt.show()
```

```python
tv_shows = df[df["type"] == "TV Show"].copy()

tv_shows["seasons"] = (
    tv_shows["duration"]
    .str.extract(r"(\d+)")
    .astype(float)
)

print(tv_shows["seasons"].value_counts().sort_index())
```

```python
plt.figure(figsize=(10, 6))

tv_shows["seasons"].value_counts().sort_index().plot(kind="bar")

plt.title("Distribution of TV Shows by Number of Seasons")
plt.xlabel("Number of Seasons")
plt.ylabel("Number of TV Shows")
plt.xticks(rotation=0)
plt.tight_layout()

plt.savefig("../outputs/tv_show_seasons_distribution.png", dpi=300)
plt.show()
```

```python
from pathlib import Path

# Create output folders
reports_dir = Path("../outputs/reports")
charts_dir = Path("../outputs/charts")

reports_dir.mkdir(parents=True, exist_ok=True)
charts_dir.mkdir(parents=True, exist_ok=True)


# 1. Content type summary
content_type_summary = (
    df["type"]
    .value_counts()
    .rename_axis("content_type")
    .reset_index(name="title_count")
)

content_type_summary["percentage"] = (
    content_type_summary["title_count"]
    / content_type_summary["title_count"].sum()
    * 100
).round(2)

content_type_summary.to_csv(
    reports_dir / "content_type_summary.csv",
    index=False
)


# 2. Rating summary
rating_summary = (
    df["rating"]
    .value_counts()
    .rename_axis("rating")
    .reset_index(name="title_count")
)

rating_summary.to_csv(
    reports_dir / "rating_summary.csv",
    index=False
)


# 3. Titles added by year
yearly_summary = (
    df["added_year"]
    .dropna()
    .value_counts()
    .sort_index()
    .rename_axis("year")
    .reset_index(name="titles_added")
)

yearly_summary.to_csv(
    reports_dir / "yearly_content_summary.csv",
    index=False
)


# 4. Country summary
country_summary = (
    country_counts
    .rename_axis("country")
    .reset_index(name="title_count")
)

country_summary.to_csv(
    reports_dir / "country_summary.csv",
    index=False
)


# 5. Genre summary
genre_summary = (
    genre_counts
    .rename_axis("genre")
    .reset_index(name="title_count")
)

genre_summary.to_csv(
    reports_dir / "genre_summary.csv",
    index=False
)


# 6. Movie duration summary
movie_duration_summary = movies["duration_minutes"].describe().round(2)

movie_duration_summary.to_csv(
    reports_dir / "movie_duration_summary.csv"
)


# 7. TV show season summary
tv_season_summary = (
    tv_shows["seasons"]
    .value_counts()
    .sort_index()
    .rename_axis("seasons")
    .reset_index(name="tv_show_count")
)

tv_season_summary.to_csv(
    reports_dir / "tv_seasons_summary.csv",
    index=False
)


# 8. Data quality summary
data_quality_summary = pd.DataFrame({
    "metric": [
        "Total rows",
        "Total columns",
        "Missing date_added",
        "Duplicate rows",
        "Duplicate show_ids"
    ],
    "value": [
        len(df),
        len(df.columns),
        int(df["date_added"].isna().sum()),
        int(df.duplicated().sum()),
        int(df["show_id"].duplicated().sum())
    ]
})

data_quality_summary.to_csv(
    reports_dir / "data_quality_summary.csv",
    index=False
)


print("EDA reports generated successfully.")
print("\nReports:")
for file in sorted(reports_dir.glob("*.csv")):
    print("-", file.name)
```
